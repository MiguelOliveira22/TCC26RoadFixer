```python
from pathlib import Path
import pandas as pd
import os

pasta_in = Path("In")
pasta_out = Path("Processed")

pasta_out.mkdir(parents=True, exist_ok=True)

arquivos_entrada = []
arquivo_saida = pasta_out / 'VolumeDiario.csv'

for dirpath, _, filename in os.walk(pasta_in):
    for file in filename:
        if "contagem_sat_dia" in file:
            arquivos_entrada.append(pasta_in / file)
```

```python
from pandas import DataFrame

volume = DataFrame()
for arquivo_entrada in arquivos_entrada:
    volume = pd.concat([volume, pd.read_csv(arquivo_entrada, encoding="utf-8-sig", dtype=str, keep_default_na=False)], ignore_index=True)

volume = volume.loc[volume.RODOVIA.str.contains("330")]
volume = volume.sort_values([volume.DATA.name], ascending=True)

volume.to_csv(arquivo_saida, index=False, encoding='utf-8-sig')
```
