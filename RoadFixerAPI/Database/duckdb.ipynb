```python
from pathlib import Path

import duckdb
import os
```

```python
FILE_PATH = Path("dados.ddb")
DATA_LOAD_PATH = Path("DuckDB")
```

```python
con = duckdb.connect(FILE_PATH)
```

```python
if not os.path.exists(DATA_LOAD_PATH):
    raise FileNotFoundError("No File or Directory Could Be Found!")

for path, _, file in os.walk(DATA_LOAD_PATH):
    con.read_csv(Path(path, file))
```
